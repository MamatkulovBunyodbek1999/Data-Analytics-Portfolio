### Importing necessary libraries  which we gonna be working with

```python
import pandas as pd
import sqlite3
```

### Lets pull csv file to df variable as pandas dataframe

```python
df=pd.read_csv("orders.csv")
df.head(2) # Lets see our dataframe
```

### Connecting sqlite to our dataframe

```python
conn = sqlite3.connect("sales.db") 
df.to_sql("orders", conn, if_exists="replace", index=False)

print("Loaded into SQLite")
```

### Lets calculate revenue by category vie sql
##### I want to see  which category is making the most revenu. To do so, basic calculations will be done via sql's SUM(), GROUP BY and ORDER BY queries


```python
query = """
SELECT Category, SUM(Sales) AS revenue
FROM orders
GROUP BY Category
ORDER BY revenue DESC
"""
# Creating a new variable result, to save our filtered data
result = pd.read_sql(query, conn)
result
```

### Now We will save our calculated revenue to a new csv file, so that our defined kpi's will be kept separately in a new file

```python
result.to_csv("revenue_by_category.csv", index=False)
```

### Down below I have  calculated revenue by product name, so that we will be able to see which products are generating the most revenue

```python
top_products = pd.read_sql("""
SELECT Product_Name, SUM(Sales) AS revenue
FROM orders
GROUP BY Product_Name
ORDER BY revenue DESC
LIMIT 10
""", conn)

top_products
```
